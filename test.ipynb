```python
d = {'a': 1, 'b': 2}

# 1. keys() : 키만 모으기
print(d.keys())    # dict_keys(['a', 'b'])

# 2. values() : 값만 모으기
print(d.values())  # dict_values([1, 2])

# 3. items() : (키, 값) 쌍 모으기
print(d.items())   # dict_items([('a', 1), ('b', 2)])

# 4. get() : 없는 키도 에러 없이 조회
print(d.get('b'))
print(d.get('c'))  # None

# 5. update() : 여러 개 한 번에 수정/추가
d.update(b=20, c=30)
print(d)           # {'a': 1, 'b': 20, 'c': 30}

# 6. pop() : 특정 키 삭제하며 값 반환
val = d.pop('a')
print(val)         # 1
print(d)           # {'b': 20, 'c': 30}

# 7. in : 존재 여부 확인 (기본은 키 검사 / .values()는 값 검사)
print('b' in d)           # True
print(20 in d)            # False (키가 아니라서)
print(20 in d.values())   # True  (값 목록에서 찾아서)
```
